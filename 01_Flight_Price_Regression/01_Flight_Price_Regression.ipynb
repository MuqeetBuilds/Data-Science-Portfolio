```python
import pandas as pd
import numpy as np
import random
np.random.seed(42)
random.seed(42)
```

```python
airlines = ['Vistara', 'Air India', 'Indigo', 'SpiceJet', 'GoFirst']
cities = ['Delhi', 'Mumbai', 'Bangalore', 'Kolkata', 'Hyderabad', 'Chennai']
times = ['Morning', 'Afternoon', 'Evening', 'Night']
```

```python
data = {
    'Airline': [random.choice(airlines) for _ in range(5000)],
    'Source_City': [random.choice(cities) for _ in range(5000)],
    'Destination_City': [random.choice(cities) for _ in range(5000)],
    'Departure_Time': [random.choice(times) for _ in range(5000)],
    'Days_Left': np.random.randint(1, 50, size=5000), # Days left until the flight
    'Duration_Hours': np.round(np.random.uniform(2.0, 15.0, size=5000), 1)
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df.loc[df['Source_City'] == df['Destination_City'],'Destination_City']='chennai'
```

```python
base_price = 3000
airline_multiplier = df['Airline'].map({'Vistara':3000,'Air India':2000,'Indigo':0,'SpiceJet':-500,'GoFirst':-500})
time_penalty = df['Days_Left'].apply(lambda x:5000 if x<5 else (2000 if x<15 else 0 ))
duration_cost = df['Duration_Hours']*500
df["price"] = base_price + airline_multiplier + time_penalty +duration_cost +np.random.randint(-1000,1000,size =5000)
print ("data successfully loaded")
```

```python
#fearure engineering
df_encoding = pd.get_dummies(df,columns = ['Airline','Source_City','Destination_City','Departure_Time'])
print(f"old shape{df.shape}")
print(f"new shape{df_encoding.shape}")
```

```python
x=df_encoding.drop('price',axis =1)
y=df_encoding['price']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print("Data successfully cleaned and split!")
print(f"Training Data (Practice Test) has {x_train.shape[0]} flights.")
print(f"Testing Data (Final Exam) has {x_test.shape[0]} flights.")
```

```python
df_encoding.shape
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error , r2_score
model = RandomForestRegressor(n_estimators=100,random_state=42)
print("Training the random forest ")
model.fit(x_train,y_train)
prediction = model.predict(x_test)
mae=mean_absolute_error(y_test,prediction)
r2=r2_score(y_test,prediction)

print("done!")
print(f"Mean absolute error : {mae}")
print(f"Accuracy Score : {r2}")
```
